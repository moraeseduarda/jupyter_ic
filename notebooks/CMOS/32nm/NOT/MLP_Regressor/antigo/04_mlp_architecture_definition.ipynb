# Multi-layer Perceptron (MPL)
Neural network for supervised learning

Idea:
- Single-output model approach (Predicting one output at a time)
- Multi-output model approach (Predicting three outputs at the same time)

## Multi-layer Perceptron for Regression using Scikit-Learn

```python
mlp_regressor= MLPRegressor(activation='relu',
                            solver = 'adam',
                            max_iter=500,
                            shuffle=True,
                            validation_fraction=0.1,
                            n_iter_no_change=10,
                            hidden_layer_sizes=(100, ),
                            early_stopping = True,
                            random_state=42,
                            verbose=True
                           )
```

- input neurons: 7 neurons (7 input variables)
- hidden layers: default 100 neurons in hidden layer
- neurons per hidden layes: 1 neuron (single-output model) or 3 neuros (multi-ouput model)
- hidden activation: relu
- output activation: relu?
- loss function: default squared_error

- activation function for hidden layer: 'relu', the rectified linear unit function, returns f(x) = max(0, x)
- solver: 'adam' refers to a stochastic gradient-based optimizer
- max_iter: 500
- shuffle: true
- validation_fraction: 0.1
- n_iter_no_change: 10
- hidden layer_sizes: 100
- early_stopping: true
- random_state: 42
- verbose: true

- learning_rate: default=’constant’
'constant' is a constant learning rate given by ‘learning_rate_init’
learning_rate_init float, default=0.001

```python
joblib.dump(mlp_regressor, "/home/eduarda/Repos/jupyter_ic/models/mlp_regressor_SKL_untrained.joblib")
```

Single-output model

```python
import tensorflow as tf

technologies = {
    "32nm_cmos": 7, # dataset number of inputs
    "7nm_finfet": 5 # dataset number of inputs
}

for tech_name, num_features in technologies.items():
    print(f"\n--- Creating and saving model for Single-Output {tech_name} ({num_features} inputs) ---")

    model = tf.keras.models.Sequential([
        tf.keras.layers.Dense(100, activation="relu", input_shape=(num_features,)), 
        tf.keras.layers.Dense(50, activation='relu'),
        tf.keras.layers.Dense(1)
    ])

    optimizer = tf.keras.optimizers.Adam(learning_rate=0.001)
    model.compile(optimizer=optimizer, loss="mean_squared_error")

    file_name = f'mlp_regressor_Keras_SingleOut_untrained_{tech_name}_inputs_{num_features}.keras'
    
    # 6. Save the compiled architecture to your directory
    full_path = f'/home/eduarda/Repos/jupyter_ic/models/{file_name}'
    model.save(full_path)
    
    print(f"Model successfully saved to: {full_path}")
```

Multi-output model

```python
import tensorflow as tf

technologies = {
    "32nm_cmos": 7, # dataset number of inputs
    "7nm_finfet": 5 # dataset number of inputs  
}

for tech_name, num_features in technologies.items():
    print(f"\n--- Creating and saving model for Multi-Output {tech_name} ({num_features} inputs) ---")

    model = tf.keras.models.Sequential([
        tf.keras.layers.Dense(100, activation="relu", input_shape=(num_features,)), 
        tf.keras.layers.Dense(50, activation='relu'),
        tf.keras.layers.Dense(3)
    ])

    optimizer = tf.keras.optimizers.Adam(learning_rate=0.001)
    model.compile(optimizer=optimizer, loss="mean_squared_error")

    file_name = f'mlp_regressor_Keras_MultiOut_untrained_{tech_name}_inputs_{num_features}.keras'
    
    # 6. Save the compiled architecture to your directory
    full_path = f'/home/eduarda/Repos/jupyter_ic/models/{file_name}'
    model.save(full_path)
    
    print(f"Model successfully saved to: {full_path}")
```
