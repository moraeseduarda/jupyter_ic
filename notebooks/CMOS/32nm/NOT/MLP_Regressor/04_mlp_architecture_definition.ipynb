# MLP Architecture Definition

```python
from sklearn.neural_network import MLPRegressor
import joblib
```

# Multi-layer Perceptron (MPL)
Neural network for supervised learning

Idea:
- Single-output model approach (Predicting one output at a time)
- Multi-output model approach (Predicting three outputs at the same time)

## Multi-layer Perceptron for Regression using Scikit-Learn

```python
mlp_regressor= MLPRegressor(activation='relu',
                            solver = 'adam',
                            max_iter=500,
                            shuffle=True,
                            validation_fraction=0.1,
                            n_iter_no_change=10,
                            early_stopping = True,
                            random_state=42,
                            verbose=False
                           )
```

- input neurons: 7 neurons (7 input variables)
- neurons per hidden layes: 1 neuron (single-output model) or 3 neuros (multi-ouput model)
- hidden activation: relu
- output activation: relu?
- loss function: default squared_error

- activation function for hidden layer: 'relu', the rectified linear unit function, returns f(x) = max(0, x)
- solver: 'adam' refers to a stochastic gradient-based optimizer
- max_iter: 500
- shuffle: true
- validation_fraction: 0.1
- n_iter_no_change: 10
- early_stopping: true
- random_state: 42
- verbose: true
